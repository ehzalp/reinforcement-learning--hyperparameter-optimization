```python
import gymnasium as gym
import minigrid
from minigrid.wrappers import ImgObsWrapper, FullyObsWrapper

# MiniGrid ortamını oluştur
env = gym.make('MiniGrid-Empty-8x8-v0', render_mode='rgb_array')
env = FullyObsWrapper(env)  # Tam gözlem (tüm ızgarayı gösterir)

print("Eylem Uzayı:", env.action_space)
print("Gözlem Uzayı:", env.observation_space)

# Rastgele bir bölüm çalıştır
obs, info = env.reset()
done = False
total_reward = 0
while not done:
    action = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated
    total_reward += reward
    env.render()
print(f"Bölüm toplam ödül: {total_reward}")
```

```python
def compute_reward(obs, done, info, prev_pos=None):
    """
    Ödül fonksiyonu.

    Args:
        obs: Gözlem sözlüğü ('image', 'direction', 'mission')
        done: Bölüm bitti mi?
        info: Ek bilgi (kullanılmayabilir)
        prev_pos: Önceki konum (x, y) - yaklaşma ödülü için

    Returns:
        reward: Hesaplanan ödül değeri
    """
    reward = 0

    # Hedefe ulaşıldıysa (done=True ve mission tamamlandıysa)
    if done and info.get('success', False):
        reward += 100.0
        return reward

    # Bölüm bitti ama hedefe ulaşılamadıysa (engel, zaman aşımı vb.)
    if done:
        reward -= 50.0
        return reward

    # TODO: Şekillendirilmiş ödül (reward shaping) ekleyin
    # İpucu: Robotun mevcut konumunu obs['image']'den veya info'dan alabilirsiniz.
    # Hedefe olan mesafe değişimine göre küçük ödül/ceza verin.
    # Her adım için küçük bir ceza (-0.1) ekleyin.

    return reward
```

```python
from gymnasium import Wrapper
import numpy as np

class CustomRewardWrapper(Wrapper):
    def __init__(self, env):
        super().__init__(env)
        self.prev_pos = None

    def reset(self, **kwargs):
        obs, info = self.env.reset(**kwargs)
        # TODO: Robotun başlangıç konumunu belirleyip self.prev_pos'a ata
        return obs, info

    def step(self, action):
        obs, reward, terminated, truncated, info = self.env.step(action)

        # TODO: compute_reward fonksiyonunu çağırarak yeni ödülü hesapla
        # custom_reward = compute_reward(...)

        # Hesaplanan ödülü reward değişkenine ata
        # reward = custom_reward

        # TODO: Yeni konumu self.prev_pos'a ata

        return obs, reward, terminated, truncated, info

# Sarılmış ortamı oluştur
base_env = gym.make('MiniGrid-Empty-8x8-v0', render_mode='rgb_array')
base_env = ImgObsWrapper(base_env) # Use ImgObsWrapper instead of FullyObsWrapper
env = CustomRewardWrapper(base_env)
```

```python
from stable_baselines3 import PPO
from stable_baselines3.common.callbacks import EvalCallback
import matplotlib.pyplot as plt

# TODO: Hiperparametreleri burada tanımlayın
config = {
    "learning_rate": 3e-4,
    "gamma": 0.99,
    "total_timesteps": 100000,
    "policy": "MlpPolicy"
}

# Modeli oluştur
model = PPO(
    config["policy"],
    env,
    learning_rate=config["learning_rate"],
    gamma=config["gamma"],
    verbose=1,
    tensorboard_log="./ppo_minigrid_tb/"
)

# Değerlendirme callback'i
eval_callback = EvalCallback(
    env,
    best_model_save_path='./logs/best_model',
    log_path='./logs/results',
    eval_freq=5000,
    deterministic=True
)

# Eğitimi başlat
model.learn(
    total_timesteps=config["total_timesteps"],
    callback=eval_callback,
    tb_log_name=f"PPO_lr{config['learning_rate']}_gamma{config['gamma']}"
)

# Modeli kaydet
model.save("ppo_minigrid_navigation")
```

```python
from stable_baselines3.common.vec_env import VecVideoRecorder, DummyVecEnv
def record_video(model, env_id, video_name='rl_navigation', video_folder='videos/'):
    def make_env():
        env = gym.make(env_id, render_mode='rgb_array')
        env = ImgObsWrapper(env) # Changed from FullyObsWrapper to ImgObsWrapper
        env = CustomRewardWrapper(env)
        return env

    vec_env = DummyVecEnv([make_env])
    vec_env = VecVideoRecorder(
        vec_env,
        video_folder,
        record_video_trigger=lambda x: x == 0,
        video_length=500,
        name_prefix=video_name
    )

    obs = vec_env.reset()
    for _ in range(200):
        action, _ = model.predict(obs, deterministic=True)
        obs, _, done, _ = vec_env.step(action)
        if done:
            break
    vec_env.close()

# Video kaydet
record_video(model, 'MiniGrid-Empty-8x8-v0')
print("Video 'videos/' klasörüne kaydedildi.")
```
